```python
from scipy.special import kolmogorov
from math import pi
import numpy as np
import multiprocess as mp
NUM_CORE = 7
pool = mp.Pool(NUM_CORE)
iter = 1000
```

```python
n = 200

# Example Sp1
Result = np.zeros(5)
for j in range(5):
    loc = j*0.05

    def Reject(i_n_loc):
        i, n, loc, kolmogorov = i_n_loc
        np.random.seed(i)
        X = np.random.normal(loc=0.0, scale=1.0, size=(n, 2)) + loc
        norms = np.sqrt(X[:,0]**2 + X[:,1]**2)
        temp = norms.argsort()
        ranks = np.empty_like(temp)
        ranks[temp] = np.arange(len(norms))
        X[:,0] /= norms
        X[:,1] /= norms
        def f(i,j):
            i, j = int(i), int(j)
            inner_pro = np.inner(X[i,:],X[j,:])
            return((inner_pro-1/4)/(17/8-inner_pro) * min(1-ranks[i]/n, 1-ranks[j]/n))
        Tn = np.sum(np.fromfunction(np.vectorize(f), (n, n), dtype=float))/n
        if kolmogorov(pi*np.sqrt(Tn)/2) <= 0.05:
            return (1)
        else:
            return (0)
        
    Result[j] = sum(pool.map(Reject, [(i, n, loc, kolmogorov) for i in range(iter)])) / iter
    print(Result[j])
```

# Henze et al., 2014

```python
# This section reproduces results for method "HHM" in Table 4 of the supplementary file.
```

```python
import numpy as np
import multiprocessing as mp

def _sphere_grid(p: int, K: int, rng: np.random.Generator) -> np.ndarray:
    """
    Returns a (K, p) array of unit vectors.
    - p=2: deterministic equally spaced angles: phi_j = 2j*pi/K, j=1..K
      (paper’s suggested choice).  :contentReference[oaicite:6]{index=6}
    - p>=3: random approximately-uniform points via normalized Gaussians.
    """
    if p == 2:
        j = np.arange(1, K + 1)
        phi = 2.0 * np.pi * j / K
        U = np.stack([np.cos(phi), np.sin(phi)], axis=1)
        return U.astype(float)

    Z = rng.normal(size=(K, p))
    norms = np.linalg.norm(Z, axis=1, keepdims=True)
    norms = np.maximum(norms, 1e-12)
    return Z / norms

def _uniform_sphere(n: int, p: int, rng: np.random.Generator) -> np.ndarray:
    """Generate (n,p) i.i.d. uniform-on-sphere directions via normalized Gaussians."""
    Z = rng.normal(size=(n, p))
    norms = np.linalg.norm(Z, axis=1, keepdims=True)
    norms = np.maximum(norms, 1e-12)
    return Z / norms

def _ks_statistic_given_grid_and_rings(X: np.ndarray, Ugrid: np.ndarray, rhos: np.ndarray) -> float:
    """
    Compute KS_{n,R} = sqrt(n) * max_{rho in rhos} max_{j,m in grid} |phi_n(rho u_j) - phi_n(rho u_m)|.
    """
    n, p = X.shape
    K = Ugrid.shape[0]

    # Precompute transpose for fast matmul
    Ugrid_T = Ugrid.T  # (p, K)

    max_over_rings = 0.0
    for rho in rhos:
        T = rho * Ugrid_T  # (p, K)
        proj = X @ T       # (n, K)
        phi = np.exp(1j * proj).mean(axis=0)  # (K,)

        # max_{j,m} |phi_j - phi_m| via pairwise distances (K up to 200 -> fine)
        diff = phi[:, None] - phi[None, :]
        ring_max = np.abs(diff).max()
        if ring_max > max_over_rings:
            max_over_rings = ring_max

    return np.sqrt(n) * max_over_rings

def ks_spherical_symmetry_test(
    X: np.ndarray,
    *,
    R: float = 2.0,
    L: int = 8,
    K: int = 9,
    alpha: float = 0.05,
    B: int = 999,
    standardize: bool = True,
    seed: int = 0,
) -> int:
    """
    KS-type test of spherical symmetry using the conditional Monte Carlo calibration:

    (i)  compute T on original data
    (ii) generate u*_j uniform on sphere
    (iii) set X*_j = u*_j * ||X_j||
    (iv) compute T* on X*
    (v) repeat B times
    (vi) reject if T > T*_(B - floor(alpha*B))

    This is exactly the resampling scheme described in the paper's Section 4

    Returns: 1 if reject, 0 otherwise.
    """
    X = np.asarray(X, dtype=float)
    if X.ndim != 2:
        raise ValueError("X must be a 2D array of shape (n, p).")
    n, p = X.shape
    if n < 2:
        raise ValueError("Need n >= 2 observations.")
    if L <= 0 or K <= 1:
        raise ValueError("Need L >= 1 and K >= 2.")
    if not (0.0 < alpha < 1.0):
        raise ValueError("alpha must be in (0,1).")
    if B <= 0:
        raise ValueError("B must be positive.")

    rng = np.random.default_rng(seed)

    # Rings: equally spaced in (0, R]
    rhos = R * (np.arange(1, L + 1) / L)

    # Build grid on S^{p-1}
    Ugrid = _sphere_grid(p=p, K=K, rng=rng)

    norms = np.linalg.norm(X, axis=1)  # ||X_j||

    # Observed test statistic
    T = _ks_statistic_given_grid_and_rings(X, Ugrid, rhos)

    # Monte Carlo calibration (conditional on norms)
    Tstars = np.empty(B, dtype=float)
    for b in range(B):
        Ustar = _uniform_sphere(n=n, p=p, rng=rng)      # u*_j
        Xstar = Ustar * norms[:, None]                  # X*_j = u*_j ||X_j||
        Tstars[b] = _ks_statistic_given_grid_and_rings(Xstar, Ugrid, rhos)

    Tstars.sort()
    # order statistic index for T*_(B - floor(alpha*B)) using 1-indexing in paper
    idx = (B - int(np.floor(alpha * B))) - 1  # convert to 0-index
    crit = Tstars[idx]

    return int(T > crit)


def ks_test_p2_8rings_9grid(X: np.ndarray, **kwargs) -> int:
    return ks_spherical_symmetry_test(X, R=2.0, L=8, K=9, **kwargs)

def ks_test_p50_8rings_200grid(X: np.ndarray, **kwargs) -> int:
    return ks_spherical_symmetry_test(X, R=2.0, L=8, K=200, **kwargs)
```

# Einmahl & Gan (JSPI 2016) Tn Test

```python
import numpy as np

def einmahl_gan_Tn_test(X, *, alpha: float = 0.05, b: float = 0.5) -> int:
    """
    Einmahl & Gan (JSPI 2016) T_n test for bivariate central symmetry about (0,0).

    Returns
    -------
    int
        1 if reject at alpha=0.05, 0 otherwise.

    Notes
    -----
    - Only alpha=0.05 is supported (uses critical value 0.5835).
    - Default bandwidth b=0.5 (as used in the paper’s simulation setup).
    """
    if abs(alpha - 0.05) > 1e-12:
        raise ValueError("This implementation only supports alpha=0.05.")
    CRIT_05 = 0.5835

    X = np.asarray(X, dtype=float)
    if X.ndim != 2 or X.shape[1] != 2:
        raise ValueError("X must be a 2D array of shape (n, 2).")
    n = X.shape[0]
    if n < 5:
        raise ValueError("n must be at least 5 for a meaningful test.")

    x = X[:, 0]
    y = X[:, 1]

    # Polar coordinates: theta_tilde in [0, 2pi)
    theta_tilde = np.mod(np.arctan2(y, x), 2.0 * np.pi)
    R = np.hypot(x, y)

    # Fold to Theta in [0, pi) and delta in {+1, -1}
    in_first_half = theta_tilde < np.pi
    Theta = np.where(in_first_half, theta_tilde, theta_tilde - np.pi)  # [0, pi)
    delta = np.where(in_first_half, 1.0, -1.0)

    # --- Compute Uhat_i = Fhat_{R|Theta}(R_i | Theta_i)
    # Rotation-invariant weights: D = 1 - cos(min(|diff|, pi-|diff|)), w = (1 - D/b)_+
    Uhat = np.empty(n, dtype=float)
    inv_b = 1.0 / b

    # internal hard-coded chunking for memory safety
    chunk = 512
    for start in range(0, n, chunk):
        end = min(n, start + chunk)
        th_block = Theta[start:end][:, None]                 # (B,1)
        d = np.abs(th_block - Theta[None, :])                # (B,n)
        d = np.minimum(d, np.pi - d)                         # circular distance on [0,pi)
        D = 1.0 - np.cos(d)
        w = 1.0 - D * inv_b
        w = np.maximum(w, 0.0)

        denom = w.sum(axis=1)                                # (B,)
        le = (R[None, :] <= R[start:end][:, None])           # (B,n)
        numer = (w * le).sum(axis=1)                         # (B,)

        Uhat[start:end] = np.divide(numer, denom, out=np.zeros_like(numer), where=denom > 0)

    # --- Compute Tn efficiently using step-function structure in u and theta empirical measure
    idx_theta = np.argsort(Theta, kind="mergesort")
    delta_sorted = delta[idx_theta]

    pos_in_theta = np.empty(n, dtype=int)
    pos_in_theta[idx_theta] = np.arange(n)

    idx_u = np.argsort(Uhat, kind="mergesort")
    U_sorted = Uhat[idx_u]

    contrib = np.zeros(n, dtype=float)  # delta-mass at each theta-rank for points with U<=current u
    total_delta = 0.0
    sqrt_n = np.sqrt(n)
    n2 = float(n * n)

    def current_f_value() -> float:
        # W at theta sample points (theta-sorted)
        W = np.cumsum(contrib) / sqrt_n
        Wpi = total_delta / sqrt_n

        prefix_W = np.cumsum(W)
        prefix_W2 = np.cumsum(W * W)
        j = np.arange(1, n + 1, dtype=float)

        # A(u) = (1/n^2) * sum_{i<=j} (W_j - W_i)^2
        SA = np.sum(j * (W * W) - 2.0 * W * prefix_W + prefix_W2)

        # B(u) = (1/n^2) * sum_{j<=i} (Wpi - W_i - W_j)^2
        a = Wpi - W
        SB = np.sum(j * (a * a) + prefix_W2 - 2.0 * a * prefix_W)

        return (SA + SB) / n2

    Tn = 0.0
    u_prev = 0.0
    ptr = 0

    for u_next in np.unique(U_sorted):
        du = float(u_next - u_prev)
        if du > 0:
            Tn += du * current_f_value()

        # add all points with U == u_next
        while ptr < n and U_sorted[ptr] <= u_next:
            k = idx_u[ptr]
            contrib[pos_in_theta[k]] += delta[k]
            total_delta += delta[k]
            ptr += 1

        u_prev = float(u_next)

    if u_prev < 1.0:
        Tn += (1.0 - u_prev) * current_f_value()

    return int(Tn > CRIT_05)
```

```python
# Example C6
iter = 1000
n = 100
rejections = np.zeros(iter)
for i in range(iter):
    np.random.seed(i)
    X = np.random.chisquare(df=1, size=(n,2)) - 1
    rejections[i] = einmahl_gan_Tn_test(X)
Result = np.mean(rejections)
print(Result)
```
